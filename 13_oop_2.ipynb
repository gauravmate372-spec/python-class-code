```python
from abc import ABC, abstractmethod
```

```python
class car(ABC):
    @abstractmethod
    def airbag(self):
        pass
```

```python
class nexon(car):
    def airbag(self):
        return'Airbags are installed in nexon'
        def button_start(self):
            return 'car will start'
```

```python
obj = nexon()
```

```python
obj.airbag()
```

```python
obj.button_start()
```

```python
class fortuner(car):
    def sunroof(self):
        return 'sunroof is opening'
```

```python
f1 = fortuner()
```

```python
class phone(ABC):
    @abstractmethod
    def call(self):
        pass
```

```python
class samsung(phone):
    def call(self):
        return 'calling from samsung phone'
    def camera(self):
            return 'opning camera from samsung mobile'
```

```python
s11 = samsung()
```

```python
s11.call()
```

```python
s11.camera()
```

```python
class Student:
    def __init__(self, name, roll_number, marks):
        self.name = name
        self.roll_number = roll_number
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll_number)
        print("Marks:", self.marks)
        print("-------------------")


# Create two objects
student1 = Student("Rahul", 101, 85)
student2 = Student("Priya", 102, 92)

# Display details
student1.display()
student2.display()
```

```python
#Employee salary
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Employee Name:", self.name)
        print("Salary:", self.salary)
        print("-------------------")


# Create three employee objects
employee1 = Employee("Amit", 30000)
employee2 = Employee("Sneha", 40000)
employee3 = Employee("Rohit", 35000)

# Display details
employee1.display()
employee2.display()
employee3.display()
```

```python
#inheritance
```

```python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, grade):
        super().__init__(name)
        self.grade = grade

    def display(self):
        print("Person Name:", self.name)
        print("Student Grade:", self.grade)


# Create student object
student = Student("Akash", "A")

# Display details
student.display()
```

```python
try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    print("\nAddition:", num1 + num2)
    print("Subtraction:", num1 - num2)
    print("Multiplication:", num1 * num2)
    print("Division:", num1 / num2)

except ZeroDivisionError:
    print("Cannot divide by zero.")

except ValueError:
    print("Please enter valid numbers.")
```

File Operations

```python
#Student data
students = """Rahul,101,85
Priya,102,92
Amit,103,78
Sneha,104,88
Rohit,105,95
"""
```

```python
#W mode
with open("student.txt", "w") as file:
    file.write(students)

print("Data written successfully using w mode.")
```

```python
#r mode
with open("student.txt", "r") as file:
    data = file.read()

print(data)
```

```python
#a mode
with open("student.txt", "a") as file:
    file.write("Neha,106,90\n")

print("New student added using a mode.")
```

```python
with open("student.txt", "r") as file:
    print(file.read())
```

```python
#r+ mode
with open("student.txt", "r+") as file:
    data = file.read()
    print("Data using r+ mode:")
    print(data)

    file.write("Kiran,107,82\n")
```

```python
#W+ mode
with open("student.txt", "w+") as file:
    file.write(students)

    file.seek(0)
    data = file.read()

    print("Data using w+ mode:")
    print(data)
```

```python
#a+ mode
with open("student.txt", "a+") as file:
    file.write("Neha,106,90\n")

    file.seek(0)
    data = file.read()

    print("Data using a+ mode:")
    print(data)
```

```python
#X mode
try:
    with open("student_new.txt", "x") as file:
        file.write(students)

    print("New file created successfully using x mode.")

except FileExistsError:
    print("File already exists.")
```

```python
#All modes
students = """Rahul,101,85
Priya,102,92
Amit,103,78
Sneha,104,88
Rohit,105,95
"""

# w mode - Write
with open("student.txt", "w") as file:
    file.write(students)

print("1. w mode - Data written successfully")


# r mode - Read
with open("student.txt", "r") as file:
    print("\n2. r mode - Reading data:")
    print(file.read())


# a mode - Append
with open("student.txt", "a") as file:
    file.write("Neha,106,90\n")

print("3. a mode - Data appended successfully")


# r+ mode - Read and Write
with open("student.txt", "r+") as file:
    print("\n4. r+ mode - Reading data:")
    print(file.read())


# w+ mode - Write and Read
with open("student.txt", "w+") as file:
    file.write(students)
    file.seek(0)

    print("\n5. w+ mode - Reading data:")
    print(file.read())


# a+ mode - Append and Read
with open("student.txt", "a+") as file:
    file.write("Neha,106,90\n")
    file.seek(0)

    print("\n6. a+ mode - Reading data:")
    print(file.read())


# x mode - Create new file
try:
    with open("student_new.txt", "x") as file:
        file.write(students)

    print("7. x mode - New file created successfully")

except FileExistsError:
    print("7. x mode - File already exists")
```

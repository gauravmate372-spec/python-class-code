```python
def decorator(func):
    def wrapper(*args,**kwargs):
        print("start")
        result = func(*args,**kwargs)
        print("End")
        return result
    return wrapper
@decorator
def add(a,b):
    return a+b
print(add(2,3))
```

```python
def gen():
    print("start")
    yield 10
    print("Middle")
    yield 20
    print("End")
g = gen()
print("A")
print(next(g))
print("B")
```

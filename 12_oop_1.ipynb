Inheritance

```python
class parent():
    def singing(self):
        return 'I am a parent class and having skill of singing'
```

```python
class child(parent):
    def cricket(self):
        return 'I am a child class and having skill of playing cricket'
```

```python
obj2 = child()
```

```python
obj2.cricket()
```

```python
obj2.singing()
```

```python
obj3 = parent()
```

```python
obj3.singing()
```

```python
class car():
    def __init__(self,engine,fuel,seat,color):
        self.engine = engine
        self.fuel = fuel
        self.seat = seat
        self.color = color
    def start(self):
        return 'car is starting'
```

```python
class bus(car):
    def off(self):
         return 'bus is getting off'
```

```python
obj4 = bus(500,'diesel',30,'white')
```

```python
obj4.start()
```

```python
obj4.off()
```

```python
obj5 = car(500,'diesel',30,'white')
```

```python
obj5.start()
```

```python
class bike(bus):
    def kickstart(self):
        return 'bike will start'
```

```python
obj6 = bike(500,'diesel',30,'white')
```

```python
obj6.kickstart()
```

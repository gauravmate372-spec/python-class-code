```python
class animal:
    def __init__(self):
        print("I am in constructor")
    def __init__(self,a,b):
        self.a=a
        self.b=b
        print(self.a+self.b)
        print(self.b)
        print("I am in 2 nd constructor")
    def eat(self):
        print("Animal eating")
a1=animal(10,20)
a1.eat() #using two constructor having a parameter
```

```python
from abc import ABC,abstractmethod
class vehicle(ABC):
    @abstractmethod
    def wheels(self):
        pass
    def run(self):
        print("I am running")
class car(vehicle):
    def wheels(self):
        print("I will run") #use the polymorphism in it override it
class bike(vehicle):
    def wheels(self):
        print("I have two wheels")
c1=car()
b1=bike()
c1.wheels()
b1.wheels()
c1.run()
b1.run()
```

```python
class account:
    def __init__(self,balance):
        self.__balanceAmt = balance
        print(self.__balanceAmt)
    def deposit(self):
        amt=int(input("Enter amount"))
        self.__balanceAmt +=amt
        print(self.__balanceAmt)
    def withdrawal(self):
        amt=int(input("Enter amount"))
        if self.__balanceAmt>=amt:
            self.__balanceAmt -=amt
            print(self.__balanceAmt)
        else:
            print("you have the insufficient balance")
c1=account(10000)
c1.deposit()
c1.withdrawal()
```
